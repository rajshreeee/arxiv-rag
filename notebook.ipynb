```python
import json

with open("data/eval_drafts.json") as f:
    drafts = json.load(f)

BATCH = 0 

VAGUE = ["the proposed", "the system", "this method", "this approach",
         "the model", "the framework", "the method"]

for d in drafts[BATCH * 10 : BATCH * 10 + 10]:
    print("=" * 80)
    print(d["qid"], "|", d["title"][:70])
    if d["copy_rate"] > 0.85:
        print("(!) high copy rate:", d["copy_rate"])
    for phrase in VAGUE:
        if phrase in d["question"].lower():
            print("(!) vague wording '", phrase, "'")
    print("\nQUESTION:", d["question"])
    print("ANSWER:  ", d["reference_answer"])
    print("\nPASSAGE: ", d["passage"][:400])
```

```python
reject = {
    "q06": "too vague; passage is figure-label debris plus a conclusion",
    "q07": "asks only for the method's name, which appears throughout the paper",
    "q22": "passage is largely synthetic benchmark/query-generation debris",
    "q38": "passage is contaminated by adversarial attack examples",
    "q54": "passage is mismatched: asks about one poster while describing another",
    "q60": "passage does not support the full causal explanation in the reference answer",
}

fixes = {
    "q04": "How does METRAG's off-the-shelf retriever score how well a document matches an input query?",
    "q05": "How does METRAG train its utility model to decide whether retrieving a document for a query is worthwhile?",
    "q10": "In a retrieval-augmented text-to-SQL system, how is the number of columns retrieved by BM25 decided from the approximated query?",
    "q14": "What technique does the paper suggest for making conflict-aware training in RAG systems more precise?",
    "q55": "How does the semantic-alignment generation framework change the role of retrieved evidence in generation?",
}
```

```python
import json

with open("data/eval_drafts.json") as f:
    drafts = json.load(f)

final = []
for d in drafts:
    if d["qid"] in reject:
        continue
    if d["qid"] in fixes:
        d["question"] = fixes[d["qid"]]
        d["was_edited"] = True
    final.append(d)

print("Final questions:", len(final))

with open("data/eval_final.json", "w") as f:
    json.dump(final, f, ensure_ascii=False, indent=1)
```
